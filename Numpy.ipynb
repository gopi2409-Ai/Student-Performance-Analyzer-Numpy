```python
import numpy as np

students = np.array([[78, 85, 90],[45, 55, 60],[92, 88, 95],[35, 40, 50],[70, 75, 80]])
subjects = ["Python", "Statistics", "NumPy"]
averages = students.mean(axis=1)
print("Student averages1:", averages)
```

```python
subject_averages = students.mean(axis=0)
print("Subject averages0:", subject_averages)
```

```python
updated_scores = np.minimum(students + 5, 100)
print(updated_scores)
```

```python
high_performers = students[averages >= 70]
print(high_performers)
```

```python
selected_students = students[[0, 2]]
print(selected_students)
```

```python
print("Highest score:", students.max())
print("Lowest score:", students.min())
print("Below 50 average:", np.sum(averages < 50))
python_high = students[students[:, 0] >= 80]
print("Python score >= 80:", python_high)
```
